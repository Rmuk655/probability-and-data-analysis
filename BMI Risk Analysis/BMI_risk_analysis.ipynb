Probability of a male with height > 160 cm has an index > 3.

```python
g = pd.read_csv("Height-Weight.csv", usecols = ["Gender", "Height", "Index"])
#Favourable number of possibilities to find the required probability (Here the number of males taller than 160 cm having an index > 3)
fav = g[(g["Height"] > 160) & (g["Index"] > 3) & (g["Gender"] == "Male")]
#Total number of possibilities (Here the number of males taller than 160 cm)
total = g[(g["Gender"] == "Male") & (g["Height"] > 160)]
p2 = len(fav)/len(total)
print(p2)
```
