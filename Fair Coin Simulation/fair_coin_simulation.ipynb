Fair Coin Simulation

Importing the necessary libraries and asking user input for bias.

```python
import numpy as np
import numpy.random as rd
import matplotlib.pyplot as plt
#Here bias is the ratio of probability of the coin landing on heads to the probability of the coin landing on tails
bias = float(input("Enter bias: "))
```

Coding the function flip_coin_biased

```python
def flip_coin_biased(bias):
    random_number = rd.rand()
    if random_number <= (bias / (bias + 1)):
        return True
    else:
        return False
```

Coding for the function fair_decision_with_biased_coin

```python
#The logic used here is that when the probability of getting a head and a tail on flipping an unfair coin are not equal, 
#the probability of getting the sequence HT and TH on flipping two such coins simultaneously are always equal to 
#the probability of a getting a head*probability of getting a tail, for these events are independent of each other.
def fair_decision_with_biased_coin(flip_coin_biased):
    while True:
        flip_1 = flip_coin_biased(bias)
        flip_2 = flip_coin_biased(bias)
        if(((flip_1 == True) and (flip_2 == False)) or ((flip_1 == False) and (flip_2 == True))):
            if(flip_1 == True):
                return True
            else:
                return False
            break
```

Performing multiple trials

```python
#Creating an empty list to append the set of first 100000 outcomes of the first function flip_coin_biased to it for plotting the PMF of the function.
outcomes_1 = []
#Creating an empty list to append the set of first 100000 outcomes of the first function fair_decision_with_biased_coin to it for plotting the PMF of the function.
outcomes_2 = []
for i in range(100000):
    out_1 = flip_coin_biased(bias)
    outcomes_1.append(out_1)
    out_2 = fair_decision_with_biased_coin(flip_coin_biased)
    outcomes_2.append(out_2)
```

Plotting the probability mass distribution function for the function flip_coin_biased using the outcomes of the trials we got previously.

```python
#The values marked on the x axis referring to the outcome given by the function flip_coin_biased.
X_axis_values_1 = [False, True]
#The frequency of the values marked on the x axis after running the function 100000 times (in above code).
freq_1 = [outcomes_1.count(False), outcomes_1.count(True)]
plt.figure(dpi = 120)
plt.title('PMF for the function flip_coin_biased')
plt.xlabel('X: Output of the function')
plt.ylabel('Y: Frequency of the output')
plt.bar(X_axis_values_1, freq_1, color='skyblue', edgecolor='black')
plt.xticks([False, True], labels=["False", "True"])
plt.show()
```

Plotting the probability mass distribution function for the function fair_decision_with_biased_coin using the outcomes of the trials we got previously.

```python
#The values marked on the x axis referring to the outcome given by the function fair_decision_with_biased_coin.
X_axis_values_2 = [False, True]
#The frequency of the values marked on the x axis after running the function 100000 times (in above code).
freq_2 = [outcomes_2.count(False), outcomes_2.count(True)]
plt.figure(dpi = 120)
plt.title('PMF for the function fair_decision_with_biased_coin')
plt.xlabel('X: Output of the function')
plt.ylabel('Y: Frequency of the output')
plt.bar(X_axis_values_2, freq_2, color ='skyblue', edgecolor ='black')
plt.xticks([False, True], labels=["False", "True"])
plt.show()
```
