Importing the necessary libraries

Probability of a person dead being of age > 30 given that the person dead is male.

```python
f = pd.read_csv("Titanic-Dataset.csv", usecols = ["Survived", "Age", "Sex"])
#Favourable number of possibilities to find the required probability (Here the number of dead males who are older than 30 years)
age_30 = f[(f["Survived"] == 0) & (f["Age"] > 30) & (f["Sex"] == "male")]
#Total number of possibilities (Here, the number of dead males)
died = f[(f["Survived"] == 0) & (f["Sex"] == "male")]
p1 = len(age_30)/len(died)
print(p1)
```
